```python
# read in csv
df_raw = pd.read_csv('data/03_Library Systembook.csv')

#  create a working copy
df = df_raw

# cols to underscore
df = pl.format_cols(df)
# drop na
df = pl.dropna(df)
# remove quotes from book checkout
df = pl.remove_quotes(df)

# correct date cols
date_cols = ['book_checkout',  'book_returned']

for c in date_cols:
    pl.format_date(df,c)

# correct id  cols
id_cols = ['id', 'customer_id']

for c in  id_cols:
    pl.format_ints(df,c)

# calculate date
df['days_borrowed'] = pl.calculate_date(df,'book_returned',  'book_checkout')
pl.format_ints(df,'days_borrowed')

# write to csv
df.to_csv('data/03_Library Systembook_cleaned.csv')
```

```python
# DF 2 transformation
# read in csv
df2_raw = pd.read_csv('data/03_Library SystemCustomers.csv')
# create a working copy
df2 = df2_raw
# format cols
df2 = pl.format_cols(df2)
# drop na
df2 = pl.dropna(df2)
# format ints
df2 = pl.format_ints(df2,'customer_id')
df2['customer_id'] = df2['customer_id'].astype("Int64")
# write to csv
df2.to_csv('data/03_Library SystemCustomers_cleaned.csv')
```
