```python
import pandas as pd
```

```python
# read in csv
df_raw = pd.read_csv('data/03_Library Systembook.csv')
df = df_raw
```

```python
#df issues
# col names should be underscore
#  date format contains   quotes
#  dates format
# null columns
# ids
```

```python
# funtions

# cols to underscore
def format_cols(df):
    df.columns = (
            df.columns
                .str.strip()
                        .str.lower()
                            .str.replace(r"\s+", "_", regex=True)
                            )
    return df

# drop na
def dropna(df):
    df = df.dropna()
    return df
    

# remove quotes from book checkout
def remove_quotes(df):
    df["book_checkout"] = (
            df["book_checkout"]
                .astype("string")
                        .str.replace('"', "", regex=False)
                            .str.strip()
                            )
    return df
# format date

def format_date(df,col):
    df[col] = pd.to_datetime(
        df[col],
            format="%d/%m/%Y",
                errors="coerce"
                )
    return df

# format ids
def format_date(df,col):
    df[col] = df[col].astype("Int64")

 # calculate date
def calculate_date(df,col1,col2):
    return (df[col1] - df[col2]).dt.days   
```

```python
# correct date cols
date_cols = ['book_checkout',  'book_returned']

for c in date_cols:
    format_date(df,c)
```

```python
# cols to underscore
df = format_cols(df)
# drop na
df = dropna(df)
# remove quotes from book checkout
df = remove_quotes(df)

# correct date cols
date_cols = ['book_checkout',  'book_returned']

for c in date_cols:
    format_date(df,c)

# # correct id  cols
# id_cols = ['id', 'customer_id']

# for c in  id_cols:
#     format_date(df,c)
```

```python
# write to csv
df.to_csv('data/03_Library Systembook_cleaned.csv')
```

```python
# DF 2 transformation
# read in csv
df2_raw = pd.read_csv('data/03_Library SystemCustomers.csv')
# create a working copy
df2 = df2_raw
# format cols
df2 = format_cols(df2)
# drop na
df2 = dropna(df2)
# format ints
df2['customer_id'] = df2['customer_id'].astype("Int64")
# write to csv
df2.to_csv('data/03_Library SystemCustomers_cleaned.csv')
```
