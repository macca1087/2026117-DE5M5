# Save a pandas DataFrame to SQLite

This notebook demonstrates a small reusable function that writes a DataFrame to a SQLite database, then reads the table back.

SQLite is included with Python through `sqlite3`; pandas provides `DataFrame.to_sql()`.

```python
import sqlite3
from pathlib import Path

import pandas as pd
```

```python
def save_dataframe_to_sqlite(df, db_path, table_name):
    """Write a pandas DataFrame to a table in a SQLite database."""
    db_path = Path(db_path)
    db_path.parent.mkdir(parents=True, exist_ok=True)

    with sqlite3.connect(db_path) as connection:
        df.to_sql(
            name=table_name,
            con=connection,
            if_exists="replace",
            index=False,
        )

    print(f"Saved {len(df)} rows to {db_path}, table '{table_name}'")
```

`if_exists="replace"` rebuilds the table each time the notebook runs, so this demo does not add duplicate rows on reruns. Use `"append"` if you want to add rows to an existing table.

```python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Ava", "Noah", "Mia"],
})

database_path = Path("library_demo.db")
table_name = "customers"

save_dataframe_to_sqlite(customers, database_path, table_name)
```

```python
with sqlite3.connect(database_path) as connection:
    saved_customers = pd.read_sql_query(
        "SELECT * FROM customers",
        connection,
    )

saved_customers
```
