# Inspect the library SQLite database

This notebook opens the database written by `library_app/app.py` and checks that both cleaned tables are present. The notebook runs from `library_app/app_data`; the database is one folder above it, beside `app.py`.

```python
import sqlite3
from pathlib import Path

import pandas as pd

database_path = (Path.cwd() / ".." / "library_app.db").resolve()
print("Database path:", database_path)
print("Database exists:", database_path.is_file())

if not database_path.is_file():
    raise FileNotFoundError(f"SQLite database not found: {database_path}")
```

```python
with sqlite3.connect(database_path) as connection:
    tables = pd.read_sql_query(
        "SELECT name FROM sqlite_master "
        "WHERE type = 'table' AND name NOT LIKE 'sqlite_%' "
        "ORDER BY name",
        connection,
    )

tables
```

```python
expected_tables = {"systembook", "customers"}
found_tables = set(tables["name"])
missing_tables = expected_tables - found_tables

if missing_tables:
    print("Missing tables:", sorted(missing_tables))
    print("Run library_app/app.py again and check its customer-processing output.")
else:
    print("Both expected tables are present.")
```

```python
with sqlite3.connect(database_path) as connection:
    customers = pd.read_sql_query("SELECT * FROM customers LIMIT 5", connection)

customers
```

```python
with sqlite3.connect(database_path) as connection:
    systembook = pd.read_sql_query("SELECT * FROM systembook LIMIT 5", connection)

systembook
```
